```python
#  # the ways of styling the grids in seaborn are:
# # white,whitegrid,dark,darkgrid,ticks
# overall style is set using set style but label and title can be set using set_context
# 4 types of context are: paper,notebook,talk,poster
# by deafault the context is set to notebook but you can change it to paper,talk,poster using set_context
# palete has 3 sets:set1,set2,set3 set1 is used for qualitative data and set2 is used for quantitative data and set3 is used for diverging data
# palette has 3 types:deep,muted,bright,dark,colorblind
sns.get_dataset_names()
penguins=sns.load_dataset('penguins')
penguins.head()
penguins['species'].value_counts()
sns.set_style('ticks')
sns.scatterplot(data=penguins,x="bill_length_mm",y="body_mass_g",hue="sex",palette="dark")
# set is used to define the type of data point like dot for male and x for female
# sns.despine(left=True,bottom=True)#by default it will remove the top and right spines but you can also remove the left and bottom spines by passing the arguments left=True and bottom=True
```

```python
sns.stripplot(data=penguins,x="species",y="body_mass_g",hue="island",palette="dark",dodge=True,jitter=True)
#in strip plot we can see the data points in a way that they overlap with each other and we cannot see the distribution of the data points
# jitter =true:jb ek hi val pr bahut saare data points hote haui toh woh ekdusre ko overlap krdete hai,jitter=true un points ke beech thoda sa RANDOM HORUZINTAL ADD krdeta hai jisse saare data points thode se fail ho jate haiaur unka count/density easily dikh jati hai
```

```python
sns.scatterplot(data=penguins,x="species",y="body_mass_g",hue="island",palette="dark")
# strip plot mein same colour ke ek sath basicaly ek hue ke ek sath aur vhi overlapping krnge
# scatterplot mein saare hues aapas mein overlap ho jate hai aur isme zyada overlapping hoti hai in comparison of strip plot
```

```python
sns.swarmplot(data=penguins,x="species",y="body_mass_g",hue="island",palette="dark")
#in swarm plot we get the data points in a way that they do not overlap with each other and we can see the distribution of the data points 
# in a better way
#the difference between strip plot and swarm plot is that in strip plot the data points overlap with each other 
# and we cannot see the distribution of the data points but in swarm plot the data points do not overlap with each other and 
# we can see the distribution of the data points in a better way
#the diff between scatter plot and swarm plot is that in scatter plot we can see the data points in a way that they overlap with each other 
# and we cannot see the distribution of the data points but in swarm plot we can see the data points in a way that they do not overlap with each other and we can see the distribution of the data points in a better way
# diff between scatter plot and strip plot is that in scatter plot we can see the data points in a way that they overlap with each other
```

```python
penguins.head()
sns.histplot(data=penguins,x="body_mass_g",stat="percent",hue="island",multiple="stack")
# sorf hue use krne se light colour aa rhe hai with kyuki layer mein overlapping hoti hai jis wajah se colours mix hote hai
# multiple="stack" is used to stack the histograms of different species on top of each other so that we can see the distribution of body mass of different species in a better way
# multiple=layer by default it is set to "layer" which means that the histograms of different species will be plotted on top of each other and we cannot see the distribution of body mass of different species in a better way
# multiple="dodge" is used to dodge the histograms of different species so that we can see the distribution of body mass of different species in a better way
```

```python
# regression plot is used to see the relationship between two variables and it is used to see the trend of the data points and it is used to see the correlation between two variables
sns.regplot(data=penguins,x="bill_length_mm",y="body_mass_g",scatter=True,fit_reg=True,ci=95)
# ci=95 is used to set the confidence interval of the regression line and by default it is set to 95% which means that the regression line will be plotted with 95% confidence interval and we can see the trend of the data points in a better way
# 95% confidence interval means that if we take 100 samples from the population and plot the regression line for each sample then 95 out of 100 regression lines will fall within the confidence interval and we can see the trend of the data points in a better way
# confidence interval is used to see the range of values in which the true value of the population parameter lies and it is used to see the uncertainty of the estimate of the population parameter and it is used to see the precision of the estimate of the population parameter
# example of confidence interval is that if we take a sample of 100 penguins and plot the regression line for the sample then the true value of the population parameter will lie within the confidence interval of the regression line with 95% confidence and we can see the trend of the data points in a better way
```
